## Trying late Binding 

```python
def outer():
    x = 10
    def inner(x=x):
        return x
    x = 20
    return inner
```

```python
x = outer()
```

```python
print(x())
```

so code executed in LEGB (Local (LOAD_FAST) => Enclosed (LOAD_DEREF) => Global (LOAD_GLOBAL) => Built In ( LOAD_GLOBAL) , functions are in latebinding that is variable are assigned values when they are executed

```python
def greet(func):
    print("I am a greet function")
    func()
    print("End of greet function")
```

```python
def check():
    print("I am check function")
```

```python
greet(check)
```

```python
@greet
def dec():
    print("I am inside a decorator")
```

```python
def decorator(func):
    def wrapper(*args,**kwargs):
        print("Indide the decorator wrapper")
        func(*args,**kwargs)
    return wrapper
```

```python
@decorator
def test(a,b):
    return a+b
```

```python
print(test(1,2))
```

```python
def decorator_function(func):
    def wrapper(*args,**kwargs):
        print("Adding some decorations right here")
        return func(*args,**kwargs)
    return wrapper
    
```

```python
@decorator_function
def tryprint():
    print("Hello world")
    
```

```python
tryprint()
```

```python
@decorator_function
def addsome(a,b):
    return a+b
```
